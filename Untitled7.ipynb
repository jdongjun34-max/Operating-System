```python
import pandas as pd
from scipy import stats
import numpy as np

df = pd.read_csv("delivery_time.csv")
x = df["delivery_min"].dropna()

mean_x = x.mean()
n = len(x)

# 표준오차
se = stats.sem(x)

# t-검정 (예: 모평균 = 0 가정, 필요하면 값 바꿔)
result = stats.ttest_1samp(x, popmean=0)

# 신뢰구간
ci_low, ci_high = stats.t.interval(
    0.95, df=n-1, loc=mean_x, scale=se
)

print("표본평균:", mean_x)
print("t-statistic:", result.statistic)
print("p-value:", result.pvalue)
print("95% CI:", (ci_low, ci_high))
```

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("before_after_worktime.csv")

before = df["before_min"]
after = df["after_min"]

data = pd.concat([before, after], axis=1).dropna()
before = data["before_min"]
after = data["after_min"]

result = stats.ttest_rel(before, after, alternative="greater")

diff = before - after

print("평균 차이(before - after):", diff.mean())
print("t-statistic:", result.statistic)
print("p-value:", result.pvalue)
```

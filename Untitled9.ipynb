````python
```python
from pathlib import Path
import pandas as pd
from scipy.stats import chi2_contingency

# =========================
# 1. 데이터 불러오기
# =========================
# Colab 기준 경로
DATA_DIR = Path('/content/type3')
file_path = DATA_DIR / 'channel_purchase.csv'

# 파일 존재 여부 확인
if not file_path.exists():
    print(f'파일을 찾을 수 없습니다: {file_path}')
else:
    # CSV 읽기
    df = pd.read_csv(file_path)

    # =========================
    # 2. 데이터 확인
    # =========================
    print('[데이터 상위 5행]')
    print(df.head(), '\n')

    print('[변수 정보]')
    df.info()
    print()

    # =========================
    # 3. 결측치 제거
    # =========================
    df = df[['channel', 'purchase_yn']].dropna()

    # purchase_yn이 문자열인 경우 숫자로 변환
    if df['purchase_yn'].dtype == 'object':
        mapping = {
            'Y': 1, 'N': 0,
            'Yes': 1, 'No': 0,
            'yes': 1, 'no': 0,
            '구매': 1, '미구매': 0
        }
        df['purchase_yn'] = df['purchase_yn'].replace(mapping)

    # 숫자로 변환
    df['purchase_yn'] = pd.to_numeric(df['purchase_yn'], errors='coerce')

    # 변환 실패한 행 제거
    df = df.dropna(subset=['purchase_yn'])

    # 정수형 변환
    df['purchase_yn'] = df['purchase_yn'].astype(int)

    # =========================
    # 4. 교차표 생성
    # =========================
    table = pd.crosstab(df['channel'], df['purchase_yn'])

    print('[교차표]')
    print(table, '\n')

    # =========================
    # 5. 카이제곱 검정
    # =========================
    chi2, p, dof, expected = chi2_contingency(table)

    print('[검정 결과]')
    print('chi-square statistic:', round(chi2, 4))
    print('p-value:', round(p, 6))
    print('degrees of freedom:', dof)

    # =========================
    # 6. 기대도수
    # =========================
    expected_df = pd.DataFrame(
        expected,
        index=table.index,
        columns=table.columns
    )

    print('\n[기대도수]')
    print(expected_df.round(2))

    # =========================
    # 7. 채널별 구매전환율
    # =========================
    conversion_rate = (
        df.groupby('channel')['purchase_yn']
        .mean()
        .sort_values(ascending=False) * 100
    )

    print('\n[채널별 구매전환율(%)]')
    print(conversion_rate.round(2))

    # =========================
    # 8. 결과 해석
    # =========================
    print('\n[해석]')
    if p < 0.05:
        print('p-value가 0.05보다 작으므로, 유입채널과 구매여부는 서로 관련이 있다고 해석할 수 있습니다.')
    else:
        print('p-value가 0.05 이상이므로, 유입채널과 구매여부의 관련성을 확인하기 어렵습니다.')

    if len(conversion_rate) > 0:
        print(
            f'가장 높은 구매전환율 채널은 {conversion_rate.index[0]}이며, '
            f'전환율은 {conversion_rate.iloc[0]:.2f}%입니다.'
        )
```
````

````python
```python
from pathlib import Path
import pandas as pd
from scipy.stats import chi2_contingency

# =========================
# 1. 데이터 불러오기
# =========================
DATA_DIR = Path('/content/type3')
file_path = DATA_DIR / 'gender_preference.csv'

# 파일 존재 여부 확인
if not file_path.exists():
    print(f'파일을 찾을 수 없습니다: {file_path}')
else:
    # CSV 읽기
    df = pd.read_csv(file_path)

    # =========================
    # 2. 데이터 확인
    # =========================
    print('[데이터 상위 5행]')
    print(df.head(), '\n')

    print('[변수 정보]')
    df.info()
    print()

    # =========================
    # 3. 필요한 열만 선택 및 결측치 제거
    # =========================
    df = df[['gender', 'content_type']].dropna()

    # 문자열 공백 제거
    df['gender'] = df['gender'].astype(str).str.strip()
    df['content_type'] = df['content_type'].astype(str).str.strip()

    # =========================
    # 4. 교차표 생성
    # =========================
    table = pd.crosstab(df['gender'], df['content_type'])

    print('[교차표]')
    print(table, '\n')

    # 교차표가 비어 있거나 행/열이 2개 미만이면 검정 불가
    if table.shape[0] < 2 or table.shape[1] < 2:
        print('카이제곱 검정을 수행하기에 데이터 범주 수가 부족합니다.')
    else:
        # =========================
        # 5. 카이제곱 검정
        # =========================
        chi2, p, dof, expected = chi2_contingency(table)

        print('[검정 결과]')
        print('chi-square statistic:', round(chi2, 4))
        print('p-value:', round(p, 6))
        print('degrees of freedom:', dof)

        # =========================
        # 6. 기대도수
        # =========================
        expected_df = pd.DataFrame(
            expected,
            index=table.index,
            columns=table.columns
        )

        print('\n[기대도수]')
        print(expected_df.round(2))

        # =========================
        # 7. 성별별 최다 선호 유형
        # =========================
        top_pref = table.idxmax(axis=1)

        print('\n[성별별 최다 선호 유형]')
        print(top_pref)

        # =========================
        # 8. 성별 내 콘텐츠 선호 비율(%)
        # =========================
        row_ratio = (
            pd.crosstab(
                df['gender'],
                df['content_type'],
                normalize='index'
            ) * 100
        )

        print('\n[성별 내 콘텐츠 선호 비율(%)]')
        print(row_ratio.round(2))

        # =========================
        # 9. 결과 해석
        # =========================
        print('\n[해석]')
        if p < 0.05:
            print('p-value가 0.05보다 작으므로, 성별에 따라 콘텐츠 선호 분포가 다르다고 해석할 수 있습니다.')
        else:
            print('p-value가 0.05 이상이므로, 성별에 따른 콘텐츠 선호 분포 차이를 확인하기 어렵습니다.')

        for g in top_pref.index:
            print(f'{g}의 최다 선호 콘텐츠는 {top_pref[g]}입니다.')
```
````
